```python
"""
Split operator module for PotentialFlow: Potential Flow Model
"""
import os
from proteus.default_so import *
from proteus import Context
import sys

# Determine the name of the file being executed
name = str(sys.argv[0][:-3])
parun_passed = False
for i in range(len(sys.argv)):
    if 'parun' in sys.argv[i]:
        parun_passed = True
    if parun_passed is True and sys.argv[i][-3:] == '.py':
        name = sys.argv[i][:-3]
        break
    else:
        name = "PotentialFlow"

# ***************************** #
# ********** CONTEXT ********** #
# ***************************** #
case = __import__(name)
Context.setFromModule(case)
ct = Context.get()

# READ FROM myPotentialFlowProblem #
assert hasattr(
    ct, 'myPotentialFlowProblem'), "Create myPotentialFlowProblem from PotentialFlowProblem"
time_stepping_scheme = ct.myPotentialFlowProblem.time_stepping_scheme
outputStepping = ct.myPotentialFlowProblem.outputStepping

# **************************** #
# ********** pnList ********** #
# **************************** #
# Define pnList based on potential flow modules (e.g., potential_p and potential_n)
pnList = [("potential_p", "potential_n")]

# **************************** #
# ********** tnList ********** #
# **************************** #
# Time stepping list for output, based on OutputStepping configuration
tnList = [0., outputStepping['dt_init']] + [
    float(k) * outputStepping['final_time'] / float(outputStepping['nDTout'])
    for k in range(1, outputStepping['nDTout'] + 1)
]
```
